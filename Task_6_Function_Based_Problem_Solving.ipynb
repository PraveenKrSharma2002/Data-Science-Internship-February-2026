#Problem Statement 1:Smart Parking Lot Management System

```python
def manage_parking(capacity, logs):
    current_vehicles = 0
    peak_usage = 0

    # Process each vehicle log
    for log in logs:
        if log == "IN":
            current_vehicles += 1
        elif log == "OUT":
            if current_vehicles > 0:
                current_vehicles -= 1

        # Track peak usage
        if current_vehicles > peak_usage:
            peak_usage = current_vehicles

    # Display current parked vehicles
    print("Currently Parked Vehicles:", current_vehicles)

    # Display parking status
    if current_vehicles > capacity:
        print("Parking Status: Capacity Exceeded")
    else:
        print("Parking Status: Available")


# Sample Input
parking_capacity = 50
vehicle_logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

manage_parking(parking_capacity, vehicle_logs)
```

#Problem Statement 2:Online Food Delivery Time Estimator

```python
def estimate_delivery_time(distance, traffic_level, weather_condition):
    # Base delivery time calculation (5 minutes per km)
    base_time = distance * 5

    # Traffic delay adjustments
    if traffic_level.lower() == "high":
        base_time += 15
    elif traffic_level.lower() == "medium":
        base_time += 8
    elif traffic_level.lower() == "low":
        base_time += 3

    # Weather delay adjustments
    if weather_condition.lower() == "rainy":
        base_time += 10
    elif weather_condition.lower() == "stormy":
        base_time += 20
    elif weather_condition.lower() == "clear":
        base_time += 0

    # Display final estimated time
    print("Estimated Delivery Time:", base_time, "minutes")


# Sample Input
distance = 8
traffic = "High"
weather = "Rainy"

estimate_delivery_time(distance, traffic, weather)
```

#Problem Statement 3:Movie Theatre Seat Occupancy Analyzer


```python
def analyze_occupancy(total_seats, booked_seats):
    # Count how many seats are booked
    booked_count = len(booked_seats)

    # Calculate occupancy percentage
    occupancy = (booked_count / total_seats) * 100

    # Print occupancy rounded to whole number
    print("Occupancy:", round(occupancy), "%")

    # Determine show status based on occupancy
    if occupancy == 100:
        print("Show Status: Housefull")
        print("Suggestion: Consider opening an additional show")

    elif occupancy >= 75:
        print("Show Status: Almost Full")

    elif occupancy >= 50:
        print("Show Status: Filling Fast")

    else:
        print("Show Status: Seats Available")


# Sample Input
total_seats = 200
booked_seats = [1] * 150   # 150 seats booked

analyze_occupancy(total_seats, booked_seats)
```

#Problem Statement 4:Cloud Server Load Classification System


```python
def classify_server_load(cpu_readings):
    # Calculate total CPU usage using a loop
    total = 0
    for reading in cpu_readings:
        total += reading

    # Calculate average CPU usage
    average = total / len(cpu_readings)

    # Print average rounded to whole number
    print("Average CPU Load:", round(average), "%")

    # Classify server status based on average
    if average < 50:
        print("Server Status: Normal")
    elif 50 <= average < 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")


# Sample Input
cpu_data = [45, 60, 70, 85, 90]

classify_server_load(cpu_data)
```

#Problem Statement 5:Smart Classroom Resource Usage Monitor


```python
def monitor_resources(resource_usage):
    # Define maximum recommended usage hours per day
    usage_limit = 8

    overused = []

    # Check each resource and its usage hours
    for resource, hours in resource_usage.items():
        if hours > usage_limit:
            overused.append(resource)

    # Display overused resources
    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")


# Sample Input
usage_data = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

monitor_resources(usage_data)
```

#Problem Statement 6: Online Event Registration Capacity Controller

```python
def manage_registrations(capacity, total_registrations):
    # Initialize counters
    confirmed = 0
    waitlisted = 0

    # Process each registration one by one
    for _ in range(total_registrations):
        if confirmed < capacity:
            confirmed += 1
        else:
            waitlisted += 1

    # Display results
    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)

    # Determine registration status
    if confirmed >= capacity:
        print("Registration Status: Closed")
    else:
        print("Registration Status: Open")


# Sample Input
event_capacity = 100
registrations = 105

manage_registrations(event_capacity, registrations)
```
