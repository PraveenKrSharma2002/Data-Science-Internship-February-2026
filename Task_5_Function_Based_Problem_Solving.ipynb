#Problem 1: Social Media – Post Engagement Analyzer 👇

```python
def analyze_post_engagement(likes_list):
  total_likes =0
  #loops to calculate total likes
  for like in likes_list:
    total_likes += like

  #conditional logic
  if total_likes >= 1000:
    status = "viral post"
  else:
    status= "Normal Engagement"

  # output
  print("Total Likes:",total_likes)
  print("Post Status:",status)

# Examples of  data  likes from different users
likes= [200, 150, 300, 100, 250, 50]
analyze_post_engagement(likes)
```

#Problem 3: Agriculture – Rainfall Adequacy Checker 👇

```python
def check_rainfall(rainfall_data, required_level):
  total_rainfall =0

  #loop to calculate total rainfall
  for rain in rainfall_data:
    total_rainfall +=rain

  #calculate average rainfall
  average_rainfall = total_rainfall / len(rainfall_data)

  #conditional logic
  if average_rainfall >= required_level:
    status = "Adequate Rainfall"
  else:
    status ="Indequate Rainfall"


  print("Avarage Rainfall:", round(average_rainfall))
  print("Rainfall status:", status)

  #example data
rainfall_list =[60,75,80,70,75,]
required_rainfall_level = 70

check_rainfall(rainfall_list, required_rainfall_level)
```

#Problem 4: Social Media – Duplicate Account Detection 👇

```python
def detect_duplicate_accounts(usernames):

  #Convert list to set
  unique_usernames = set(usernames)

  # check for duplicates
  if len(unique_usernames) < len(usernames):
    result = "yes"
  else:
    result= " no"

  print("Duplicates Accounts found:",result)

#Example data
users=["khushboo2000","praveen121","kumar212","khushboo2000","acharya786"]
detect_duplicate_accounts(users)
```

#Problem 6: Agriculture – Premium Crop Price Filter 👇

```python
def filter_premium_crops(price_list):

    premium_crops = []

    # Loop + Filtering Logic
    for price in price_list:
        if price > 2000:
            premium_crops.append(price)

    # Output
    print("Premium Crops:", premium_crops)


# Example crop prices (₹)
crop_prices = [1500, 2500, 1800, 3200, 1900]

filter_premium_crops(crop_prices)
```

#Problem 7: System Monitoring – Application Health Checker 👇

```python
def check_application_health(error_count):

    # Conditional Logic
    if error_count == 0:
        status = "Healthy"
    elif error_count <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"

    # Output
    print("Error Count:", error_count)
    print("System Status:", status)


# Example
check_application_health(7)
```

#Problem 9: E-Learning – Student Attendance Eligibility System 👇


```python
def check_attendance_eligibility(attendance_list):

    total_classes = len(attendance_list)
    attended_classes = 0

    # Loop to count attended classes
    for status in attendance_list:
        if status == 1:   # 1 means present
            attended_classes += 1

    # Calculate attendance percentage
    attendance_percentage = (attended_classes / total_classes) * 100

    # Check eligibility
    if attendance_percentage >= 75:
        eligibility = "Eligible"
    else:
        eligibility = "Not Eligible"

    # Output
    print("Attendance Percentage:", round(attendance_percentage, 1))
    print("Exam Eligibility:", eligibility)


# Example data (1 = Present, 0 = Absent)
attendance_data = [1, 1, 1, 0, 1]  # 4 out of 5 present → 80%

check_attendance_eligibility(attendance_data)
```

#Problem 10: Smart Electricity Bill Analyzer 👇


```python
def calculate_electricity_bill(units):

    # Bill Calculation using slab logic
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)

    # Usage Classification
    if bill < 500:
        status = "Low Usage"
    elif 500 <= bill <= 1500:
        status = "Moderate Usage"
    else:
        status = "High Usage"

    return bill, status


# Example
units_consumed = 250
total_bill, usage_status = calculate_electricity_bill(units_consumed)

print("Units Consumed:", units_consumed)
print("Total Bill: ₹", total_bill)
print("Usage Level:", usage_status)
```
