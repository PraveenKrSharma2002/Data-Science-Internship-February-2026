#Problem 1: Mobile Recharge Validation System

```python
def recharge():
  # list of valid recharge plans
  plans = [199,299,399,599]

  # loops runs until user enters a valid amount
  while True:
    # take recharge amount input from user
    amount = int(input("enter recharge Amount: ₹"))

    #check minimum recharge rule
    if amount < 50:
      print("recharge must be at least ₹50")

    #check if amount matches available plans
    elif amount not in plans:
      print("Invalid plan, Availables plans: ",plans)

    else:
      #valid recharge
      print("Recharge successful")
      break #exit loops after success

    #message shown when input is invalid
    print("Try again\n")

#function call
recharge()
```

#Problem 2: Inventory Reorder Alert System

```python
def check_inventory(stock):
    # loop through each product and its quantity
    for product, quantity in stock.items():

        # check if stock is below minimum level
        if quantity < 15:
            print(product, ": Reorder Alert")

        else:
            print(product, ": Stock OK")


# sample inventory data
inventory = {
    "Rice": 10,
    "Wheat": 25,
    "Sugar": 12,
    "Oil": 18
}

# function call
check_inventory(inventory)
```

#Problem 3: Student Result Processing System

```python
def check_result(marks):
    # calculate total marks using loop
    total = 0
    for m in marks:
        total += m

    # find average
    average = total / len(marks)

    # display average
    print("Average Marks:", average)

    # check pass or fail condition
    if average >= 50:
        return "Pass"
    else:
        return "Fail"


# sample student marks
student_marks = [60, 55, 70, 40]

# function call
result = check_result(student_marks)
print("Result:", result)
```

#Problem 4: Cab Fare Estimator with Retry Option

```python
def calculate_fare(distance, peak):
    base_fare = 50          # fixed starting fare
    per_km_rate = 12        # cost per kilometer

    # calculate basic fare
    fare = base_fare + (distance * per_km_rate)

    # add 25% extra during peak hours
    if peak.lower() == "yes":
        fare = fare + (fare * 0.25)

    return fare


def cab_fare_system():
    while True:
        # take distance input
        km = float(input("Enter distance (in km): "))

        # ask if it is peak hour
        peak_hour = input("Is it peak hour? (yes/no): ")

        # calculate fare
        total_fare = calculate_fare(km, peak_hour)

        print("Estimated Fare: ₹", round(total_fare, 2))

        # ask user if they want to calculate again
        retry = input("Do you want to calculate again? (yes/no): ")
        if retry.lower() != "yes":
            print("Thank you for using fare estimator.")
            break   # exit loop


# function call
cab_fare_system()
```

#Problem 5: Employee Attendance Eligibility Checker

```python
def check_eligibility(attendance):
    present_days = 0   # counter for present days

    # count number of present days using loop
    for day in attendance:
        if day == "P":
            present_days += 1

    # calculate attendance percentage
    percentage = (present_days / len(attendance)) * 100
    print("Attendance Percentage:", percentage)

    # check eligibility condition
    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"


# sample attendance record
record = ["P", "P", "A", "P", "P", "A", "P", "P"]

# function call
status = check_eligibility(record)
print("Status:", status)
```

#Problem 6: Password Strength Checker

```python
def check_password_strength(password):
    # rule flags
    has_digit = False
    has_special = False

    # check each character in password
    for char in password:
        if char.isdigit():              # check for number
            has_digit = True
        elif char in "@#$":             # check for special character
            has_special = True

    # check all conditions
    if len(password) < 8:
        return "Weak,Minimum length is 8"

    if not has_digit:
        return "Weak, Add at least one digit"

    if not has_special:
        return "Weak, Add at least one special character (@#$..)"

    return "Strong Password"


# take password input
user_password = input("Enter password: ")

# function call
result = check_password_strength(user_password)
print(result)
```
